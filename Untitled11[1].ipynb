```python
import sqlite3
from pathlib import Path

DB_PATH = Path("/content/iot.db")

def init_db():
    conn = sqlite3.connect(DB_PATH)
    cur = conn.cursor()
    cur.execute("""
    CREATE TABLE IF NOT EXISTS sensor_data (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        device_id TEXT NOT NULL,
        ts DATETIME DEFAULT CURRENT_TIMESTAMP,
        temperature REAL,
        humidity REAL
    );
    """)
    cur.execute("CREATE INDEX IF NOT EXISTS idx_ts ON sensor_data(ts);")
    cur.execute("CREATE INDEX IF NOT EXISTS idx_device_ts ON sensor_data(device_id, ts);")
    conn.commit()
    conn.close()

init_db()
print("DB ready at:", DB_PATH)
```

```python
import threading
from flask import Flask, request, jsonify
import sqlite3
from pathlib import Path

DB_PATH = Path("/content/iot.db")  # если у тебя в ячейке 2 другой путь — поставь тот же

app = Flask(__name__)  # <-- ВОТ ТУТ ВАЖНО: name

@app.get("/health")
def health():
    return jsonify({"status": "ok"})

@app.post("/ingest")
def ingest():
    data = request.get_json(force=True)

    device_id = data.get("device_id", "unknown")
    temperature = data.get("temperature")
    humidity = data.get("humidity")

    conn = sqlite3.connect(DB_PATH)
    cur = conn.cursor()
    cur.execute(
        "INSERT INTO sensor_data (device_id, temperature, humidity) VALUES (?, ?, ?)",
        (device_id, temperature, humidity)
    )
    conn.commit()
    conn.close()

    return jsonify({"status": "saved", "data": data})

def run_server():
    app.run(host="127.0.0.1", port=5000, debug=False, use_reloader=False)

threading.Thread(target=run_server, daemon=True).start()
print("Server started on http://127.0.0.1:5000")
```

```python
import time, math, random, requests

BASE = "http://127.0.0.1:5000"
URL = f"{BASE}/ingest"

DEVICE_ID = "sim1"

def send_points(n=60, period_sec=2):
    t0 = time.time()
    ok = 0
    for i in range(n):
        t = time.time() - t0
        temperature = 22 + 2*math.sin(t/10) + random.uniform(-0.2, 0.2)
        humidity = 45 + 5*math.sin(t/12) + random.uniform(-0.5, 0.5)

        payload = {
            "device_id": DEVICE_ID,
            "temperature": round(temperature, 2),
            "humidity": round(humidity, 2)
        }

        try:
            r = requests.post(URL, json=payload, timeout=5)
            if r.status_code == 200:
                ok += 1
            print(f"{i+1}/{n} sent:", payload, "|", r.status_code)
        except Exception as e:
            print("error:", e)

        time.sleep(period_sec)

    print(f"Done. Successful: {ok}/{n}")

send_points(n=60, period_sec=2)
```

```python
import sqlite3
import pandas as pd
import plotly.express as px

DB_PATH = "/content/iot.db"

# сколько минут показать (можешь менять)
MINUTES = 60

conn = sqlite3.connect(DB_PATH)
df = pd.read_sql_query(f"""
SELECT ts, device_id, temperature, humidity
FROM sensor_data
WHERE ts >= datetime('now', '-{MINUTES} minutes')
ORDER BY ts
""", conn)
conn.close()

df["ts"] = pd.to_datetime(df["ts"])

if df.empty:
    print("Нет данных за выбранный период. Увеличь MINUTES или запусти симулятор ещё раз.")
else:
    last = df.iloc[-1]
    print(f"Показываю последние {MINUTES} минут. Всего точек: {len(df)}")
    print("Последнее значение:", last.to_dict())

    # метрики (просто выводом)
    print(f"Last temp: {last['temperature']} °C")
    print(f"Last humidity: {last['humidity']} %")

    # график температуры
    fig_t = px.line(df, x="ts", y="temperature", color="device_id", title="Temperature over time")
    fig_t.show()

    # график влажности
    fig_h = px.line(df, x="ts", y="humidity", color="device_id", title="Humidity over time")
    fig_h.show()

    # показать таблицу последних точек
    display(df.tail(20))
```

```python
from IPython.display import clear_output
import time
import sqlite3
import pandas as pd
import plotly.express as px

DB_PATH = "/content/iot.db"

REFRESH_SECONDS = 2
UPDATES = 15          # сколько раз обновлять
MINUTES = 10          # последние N минут показывать

for i in range(UPDATES):
    conn = sqlite3.connect(DB_PATH)
    df = pd.read_sql_query(f"""
    SELECT ts, device_id, temperature, humidity
    FROM sensor_data
    WHERE ts >= datetime('now', '-{MINUTES} minutes')
    ORDER BY ts
    """, conn)
    conn.close()

    df["ts"] = pd.to_datetime(df["ts"])

    clear_output(wait=True)
    print(f"Live dashboard update {i+1}/{UPDATES} | points: {len(df)}")

    if not df.empty:
        last = df.iloc[-1]
        print("Last:", last.to_dict())

        fig = px.line(df, x="ts", y="temperature", color="device_id", title="Live Temperature (last minutes)")
        fig.show()
    else:
        print("Нет данных. Запусти симулятор (ячейка 5).")

    time.sleep(REFRESH_SECONDS)
```
